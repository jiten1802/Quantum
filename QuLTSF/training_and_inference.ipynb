### Baseline Reproduction 

```python
configs = QuLTSF_Base_Config()
train_loader, val_loader, test_loader, data_dict = get_all_loaders(configs)
```

```python
qultsf_baseline = QuLTSF_Model(configs)

run_training_pipeline(
    model=qultsf_baseline,
    train_loader=train_loader,
    val_loader=val_loader,
    scaler=data_dict['scaler'],
    experiment_name="weather_qultsf_baseline"
)
```
